```python
#Exploring the Structure of Sentinel-5P NetCDF Files
```

```python
from netCDF4 import Dataset

file_path = r"C:\Users\intel\Downloads\S5P_NRTI_L2__NO2____20241020T085816_20241020T090316_36376_03_020701_20241020T093206.nc"

def explore_group(group, indent=0):
    space = " " * indent
    print(f"{space}Group: {group.path}")
    
   
    for var_name, var in group.variables.items():
        units = var.units if hasattr(var, 'units') else 'No units'
        print(f"{space}  Variable: {var_name}")
        print(f"{space}    Dimensions: {var.dimensions}")
        print(f"{space}    Shape: {var.shape}")
        print(f"{space}    Units: {units}")
    
    
    for sub_group_name, sub_group in group.groups.items():
        explore_group(sub_group, indent + 2)


with Dataset(file_path, mode='r') as nc_file:
    explore_group(nc_file)
```

```python
#Listing Key Variables in Sentinel-5P NO₂ NetCDF Files
```

```python
from netCDF4 import Dataset

file_path = r"C:\Users\intel\Downloads\S5P_NRTI_L2__NO2____20241020T085816_20241020T090316_36376_03_020701_20241020T093206.nc"

with Dataset(file_path, mode='r') as nc:
    print(" File successfully loaded!")
    print("=" * 80)
    print(f"File: {file_path}")
    print("=" * 80)

    if "PRODUCT" in nc.groups:
        main_group = nc.groups["PRODUCT"]
        print("\n📂 Main group found: 'PRODUCT'")
    else:
        main_group = nc 
        print("\n⚠️ No 'PRODUCT' group found — showing top-level variables instead.")

    print("\n📊 Main Variables:")
    print("-" * 80)

    for var_name, var in main_group.variables.items():
        units = getattr(var, "units", "—")
        long_name = getattr(var, "long_name", "—")
        shape = var.shape
        print(f"{var_name:45} | Units: {units:15} | Shape: {shape} | Description: {long_name}")

    print("\n✅ Summary complete.")
```
