```python
import geopandas as gpd
import matplotlib.pyplot as plt

# Shapefile Path
shapefile_path = r"D:\Education\University Files\Masters\Research\Master’s Thesis\Data\Shape\Tehran.json"

# Reading a Shapefile
gdf = gpd.read_file(shapefile_path)

# Print General Information
print(gdf.head())
print ("\n")
print(f"Projection (CRS): {gdf.crs}")
print ("\n")
# Plotting
gdf.plot(edgecolor="black", facecolor="lightblue", figsize=(8, 8))
plt.title("Shapefile Preview", fontsize=14)
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```
