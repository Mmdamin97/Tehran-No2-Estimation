```python
# Here’s a Python code that opens a NetCDF file, prints each variable’s min, max, and units.
```

```python
import netCDF4 as nc
import numpy as np

file_path = r"C:\Users\intel\Downloads\data_sfc.nc"

# Open and inspect variables
dataset = nc.Dataset(file_path)

print(f"\n📘 File: {file_path}")
print(f"📊 Variables found: {list(dataset.variables.keys())}\n")

for var_name, var_obj in dataset.variables.items():
    try:
        # Load variable data
        data = var_obj[:]
        
        # Skip non-numeric variables
        if not hasattr(data, "dtype") or not np.issubdtype(data.dtype, np.number):
            print(f"🔸 {var_name}: non-numeric or empty variable, skipped.")
            continue
        
        # Compute min & max
        vmin = np.nanmin(data)
        vmax = np.nanmax(data)
        
        # Try to get metadata
        units = getattr(var_obj, "units", "—")
        long_name = getattr(var_obj, "long_name", "—")

        print(f"🔹 {var_name}:")
        print(f"    📏 Range: {vmin:.6f} → {vmax:.6f}")
        print(f"    🧾 Units: {units}")
        print(f"    🏷️  Description: {long_name}\n")

    except Exception as e:
        print(f"⚠️ {var_name}: could not process ({e})")

dataset.close()
```
