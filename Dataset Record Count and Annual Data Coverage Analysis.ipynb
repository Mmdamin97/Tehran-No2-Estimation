```python
#Uses Pandas and OS to read and process Excel files (.xlsx and .xls) from a specified folder.
#Iterates through all Excel files in the directory:
#- Reads .xlsx files using the openpyxl engine.
#- Reads .xls files using the xlrd engine.
#- Calculates and prints the number of rows in each successfully read file.
#Tracks the total number of rows and the number of readable Excel files.
#Skips files that cannot be read and prints a warning message.
#Calculates the following statistics:
#- Average number of rows per readable file by dividing the total rows by the number of readable files.
#- Percentage of annual data coverage by dividing the average number of rows per file by 365 and multiplying by 100.
#Prints the final summary, including total records, readable files, average records per file, and estimated percentage of the year covered by the data.
```

```python
import pandas as pd
import os

folder_path = r"C:\Users\intel\Documents\Environmental\2024\Located , Averaged , Converted, DJF Format"  # update your folder path

total_rows = 0
readable_files = 0  # count of successfully read files

for file in os.listdir(folder_path):
    if file.endswith('.xlsx') or file.endswith('.xls'):
        file_path = os.path.join(folder_path, file)
        
        try:
            # Try to read the Excel file
            if file.endswith('.xlsx'):
                df = pd.read_excel(file_path, engine='openpyxl')
            else:
                df = pd.read_excel(file_path, engine='xlrd')
            
            num_rows = len(df)
            print(f"{file}: {num_rows} rows")
            total_rows += num_rows
            readable_files += 1
        
        except Exception as e:
            # Skip files that cannot be read
            print(f"⚠️ Could not read {file}: {e}")

# Calculate average rows per readable file
if readable_files > 0:
    average_rows = total_rows / readable_files
    percentage_of_year = (average_rows / 365) * 100  # percentage of year with data
else:
    average_rows = 0
    percentage_of_year = 0

print(f"\nTotal number of rows in all readable Excel files: {total_rows}")
print(f"Number of readable Excel files: {readable_files}")
print(f"Average number of rows per readable file: {average_rows:.2f}")
print(f"Percentage of the year covered by data: {percentage_of_year:.2f}%")
```
