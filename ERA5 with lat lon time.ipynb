```python
import xarray as xr

file_path = r"C:\Users\intel\Documents\ERA5\9b73e070aace4e5f3b780c7f278e85c3\data_stream-oper_stepType-accum.nc"

dataset = xr.open_dataset(file_path)

print("General information about the file:")
print(dataset)


print("\nVariables in the file:")
print(list(dataset.data_vars))  # List of Variables


if 'latitude' in dataset and 'longitude' in dataset:
    dataset['lat'] = dataset['latitude']  
    dataset['lon'] = dataset['longitude']  


if 'time' in dataset:
    dataset['time_var'] = dataset['time']  


df = dataset.to_dataframe().reset_index()  


print("\nTable of variables and values ​​(including longitude, latitude and time):")
print(df.head())  

df.to_csv("netcdf_variables_with_time_lat_lon.csv", index=False)
print("\nThe table was saved in the file netcdf_variables_with_time_lat_lon.csv..")
```
