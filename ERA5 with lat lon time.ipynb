```python
import xarray as xr

file_path = "C:/Users/intel/Documents/Thesis Data/ERA5/1114f8ac963e6cc3de6f8571670bbe55/data_stream-oper_stepType-instant.nc"

dataset = xr.open_dataset(file_path)


print("اطلاعات کلی درباره فایل:")
print(dataset)


print("\nمتغیرهای موجود در فایل:")
print(list(dataset.data_vars))  # List of Variable


if 'latitude' in dataset and 'longitude' in dataset:
    dataset['lat'] = dataset['latitude']  
    dataset['lon'] = dataset['longitude']  


if 'time' in dataset:
    dataset['time_var'] = dataset['time']  


df = dataset.to_dataframe().reset_index()  


print("\nجدول متغیرها و مقادیر (شامل طول، عرض جغرافیایی و زمان):")
print(df.head())  

df.to_csv("netcdf_variables_with_time_lat_lon.csv", index=False)
print("\nجدول در فایل netcdf_variables_with_time_lat_lon.csv ذخیره شد.")
```
